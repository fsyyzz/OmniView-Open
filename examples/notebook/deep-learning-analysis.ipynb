# 🚀 深度学习模型收敛性与度量分析实验
本文档展示使用 PyTorch 训练 Transformer 编码器过程中的损失函数收敛曲线与模型评估度量。

$$\mathcal{L}_{\text{total}} = \frac{1}{N} \sum_{i=1}^N \ell(y_i, \hat{y}_i) + \lambda \|\Theta\|_2^2$$


```python
import torch
import torch.nn as nn
import numpy as np

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print(f"Using compute backend: {device}")
```

```python
# 模拟 5 个 Epoch 的验证集损失收敛度量
epochs = 5
for epoch in range(1, epochs + 1):
    loss = 0.85 * np.exp(-0.4 * epoch) + 0.05
    acc = 0.70 + 0.25 * (1 - np.exp(-0.5 * epoch))
    print(f"Epoch [{epoch}/{epochs}] - Loss: {loss:.4f} | Val Accuracy: {acc*100:.2f}%")
```

```python
# 绘制模型收敛图表与混淆矩阵
import matplotlib.pyplot as plt
plt.figure(figsize=(8, 4.5))
plt.title("Convergence Curve")
plt.show()
```
